# Clean Image Generator
Generates the needed files for clean, high-quality images.

```python
import sys
import os

sys.path.append(os.path.abspath(os.path.join(os.getcwd(), "..")))

import pandas as pd
import json

from scripts.data_loader import generate_target_dataset
```

```python
# full dataset
full_dataset = generate_target_dataset(
    "../data/caption-dataset/annotations/train.json",
    "../data/image-quality-assessment/annotations/train.json",
)
print(f"Full dataset length: {len(full_dataset)}")
print(json.dumps(full_dataset[0], indent=4))
```

```python
# load filter dataset
all_filter_data = pd.concat(
    [
        pd.read_csv("./input-data/high-quality_images-rating-4.csv"),
        pd.read_csv("./input-data/high-quality_images-rating-5.csv"),
    ]
)
all_filter_data = all_filter_data[all_filter_data["product"] != "Y"]
display(all_filter_data.head())

# get relevant ids
relevant_ids = all_filter_data["image_id"].unique()
print(f"Number of relevant ids: {len(relevant_ids)}")
print(relevant_ids[0:10])
```

```python
# filter full dataset using relevant ids
filtered_dataset = list(
    filter(lambda x: x["image_id"] in set(relevant_ids), full_dataset)
)
print(f"Filtered dataset length: {len(filtered_dataset)}")

# save filtered dataset
os.makedirs(
    os.path.dirname("../data/study-2-input/high-quality-images/"), exist_ok=True
)
with open(
    "../data/study-2-input/high-quality-images/high-quality-images-filtered.json",
    "w",
    encoding="utf-8",
) as f:
    json.dump(filtered_dataset, f, indent=4)
```

```python
# show summary statistics for each image quality column
filtered_dataset_df = pd.DataFrame.from_dict(filtered_dataset)

# combine all this into a single table
# Create a summary table with value counts for each image quality column
quality_columns = [
    "unrecognizable",
    "blur",
    "framing",
    "obstruction",
    "rotation",
    "too dark",
    "too bright",
    "other",
    "no issue",
]

summary_table = (
    pd.DataFrame(
        {
            col: filtered_dataset_df[col].value_counts().sort_index()
            for col in quality_columns
        }
    )
    .fillna(0)
    .astype(int)
)

display(summary_table)
```
