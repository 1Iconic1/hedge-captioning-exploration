```python
from sentence_transformers import SentenceTransformer
```

```python
attributes1 = [
    "text exact match",
    "design",
    "florets de brocoli",
    "daiya mozzarella style shred",
    "box",
    "box",
    "box",
    "dog white",
    "black",
    "right",
]
attributes2 = [
    "text exact match",
    "cow",
    "fleurets de broccoli",
    "daiya mozzarella style shreds",
    "package",
    "carton",
    "container",
    "dog black",
    "white",
    "wrong",
]
```

```python
attributes2 = [
    "Carton",
    "Dawn Smooth n",
    "orange juice",
    "dairy goodness",
    "packaging",
    "top",
    "bottom",
    "image",
    "slice",
]

attributes1 = ["Carton"] * len(attributes2)
```

```python
def compute_similarity(model, attributes):
    """Computes pairwise similarity between all attributes in the list.

    Args:
        model (HuggingFace SentenceTransformer): SentenceTransformer model
        attributes (list): List of attributes to compute similarity for
    """

    # Compute embeddings for both lists
    embeddings = model.encode(attributes)

    # Compute cosine similarities
    similarities = model.similarity(embeddings, embeddings)
    print(similarities)


def compute_pairwise_similarity(model, attributes1, attributes2):
    assert len(attributes1) == len(attributes2)

    # Compute embeddings for both lists
    embeddings1 = model.encode(attributes1)
    embeddings2 = model.encode(attributes2)

    # Compute cosine similarities
    similarities = model.similarity_pairwise(embeddings1, embeddings2)

    # Output the pairs with their score
    for idx_i, attribute1 in enumerate(attributes1):
        print(attribute1)
        attribute2 = attributes2[idx_i]
        print(f" - {attribute2: <30}: {similarities[idx_i]:.4f}")
```

```python
compute_similarity(SentenceTransformer("all-mpnet-base-v2", device="mps"), attributes2)
```

```python
# all-mpnet-base-v2: general purpose model (https://huggingface.co/sentence-transformers/all-mpnet-base-v2)
compute_pairwise_similarity(
    SentenceTransformer("all-mpnet-base-v2", device="mps"), attributes1, attributes2
)
```

```python
# all-mpnet-base-v2: general purpose model (https://huggingface.co/sentence-transformers/all-mpnet-base-v2)
compute_pairwise_similarity(
    SentenceTransformer("Qwen/Qwen3-Embedding-8B", device="mps"),
    attributes1,
    attributes2,
)
```

```python
# all-mpnet-base-v2: general purpose model (https://huggingface.co/sentence-transformers/all-mpnet-base-v2)
compute_pairwise_similarity(
    SentenceTransformer("microsoft/mpnet-base", device="mps"),
    attributes1,
    attributes2,
)
```

```python
# microsoft/deberta-xlarge-mnli: highest correlation with human ratings, but tuned for NLI:
compute_pairwise_similarity(
    SentenceTransformer("microsoft/deberta-xlarge-mnli", device="mps"),
    attributes1,
    attributes2,
)
```

```python
# distiluse-base-multilingual-cased-v1: multilingual semantic similarity
compute_pairwise_similarity(
    SentenceTransformer("distiluse-base-multilingual-cased-v1"),
    attributes1,
    attributes2,
)
```

```python
# google/embeddinggemma-300m: google's new general-purpose embedding model, which (supposedly) powers the Gemma and Gemini models (https://huggingface.co/google/embeddinggemma-300m)
compute_pairwise_similarity(
    SentenceTransformer("google/embeddinggemma-300m"), attributes1, attributes2
)
```
